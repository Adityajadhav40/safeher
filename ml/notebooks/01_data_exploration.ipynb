```python
# Check whether State + District + Year uniquely identifies each record
# and whether the IPC and women-crime datasets have matching keys.

historical_ipc = primary_datasets["ipc_2001_2013"]
historical_women = primary_datasets["women_2001_2013"]

# Filter Maharashtra
ipc_mh = historical_ipc[
    historical_ipc["State_UT"].astype(str).str.strip().str.upper() == "MAHARASHTRA"
].copy()

women_mh = historical_women[
    historical_women["State_UT"].astype(str).str.strip().str.upper() == "MAHARASHTRA"
].copy()

# Define the join key
key_columns = ["State_UT", "District", "Year"]

# Check duplicate keys
ipc_duplicates = ipc_mh.duplicated(subset=key_columns).sum()
women_duplicates = women_mh.duplicated(subset=key_columns).sum()

print("IPC duplicate keys:", ipc_duplicates)
print("Women-crime duplicate keys:", women_duplicates)

# Compare keys between the two datasets
ipc_keys = set(map(tuple, ipc_mh[key_columns].itertuples(index=False, name=None)))
women_keys = set(map(tuple, women_mh[key_columns].itertuples(index=False, name=None)))

print("\nIPC unique keys:", len(ipc_keys))
print("Women-crime unique keys:", len(women_keys))

print("\nKeys present in IPC but missing in women-crime:", len(ipc_keys - women_keys))
print("Keys present in women-crime but missing in IPC:", len(women_keys - ipc_keys))

if ipc_keys - women_keys:
    print("\nExamples missing from women-crime:")
    print(list(ipc_keys - women_keys)[:10])

if women_keys - ipc_keys:
    print("\nExamples missing from IPC:")
    print(list(women_keys - ipc_keys)[:10])
```

```python
# Compare Maharashtra district names between historical data and 2014

ipc_2014 = primary_datasets["ipc_2014"]
women_2014 = primary_datasets["women_2014"]

# Filter Maharashtra 2014
ipc_2014_mh = ipc_2014[
    ipc_2014["State_UT"].astype(str).str.strip().str.upper() == "MAHARASHTRA"
].copy()

women_2014_mh = women_2014[
    women_2014["State_UT"].astype(str).str.strip().str.upper() == "MAHARASHTRA"
].copy()

historical_districts = set(ipc_mh["District"].dropna().astype(str).str.strip())
ipc_2014_districts = set(ipc_2014_mh["District"].dropna().astype(str).str.strip())
women_2014_districts = set(women_2014_mh["District"].dropna().astype(str).str.strip())

print("Historical districts:", len(historical_districts))
print("IPC 2014 districts:", len(ipc_2014_districts))
print("Women 2014 districts:", len(women_2014_districts))

print("\n" + "=" * 80)
print("Historical → IPC 2014 differences")
print("=" * 80)

print("\nPresent historically but not in 2014:")
print(sorted(historical_districts - ipc_2014_districts))

print("\nPresent in 2014 but not historically:")
print(sorted(ipc_2014_districts - historical_districts))
```

```python
# Normalize district names to a common representation

def normalize_district_name(value):
    """
    Convert district names from different source formats
    into a common comparable representation.
    """
    if pd.isna(value):
        return None
    
    name = str(value).strip().upper()

    # Known naming differences between historical and 2014 data
    district_mapping = {
        "MUMBAI RLY.": "MUMBAI RAILWAY",
        "MUMBAI RAILWAY": "MUMBAI RAILWAY",
        
        "NAGPUR RLY.": "NAGPUR RAILWAY",
        "NAGPUR RAILWAY": "NAGPUR RAILWAY",
        
        "PUNE RLY.": "PUNE RAILWAY",
        "PUNE RAILWAY": "PUNE RAILWAY",
    }

    return district_mapping.get(name, name)


# Test normalization on all Maharashtra district names
historical_normalized = {
    normalize_district_name(x)
    for x in historical_districts
}

ipc_2014_normalized = {
    normalize_district_name(x)
    for x in ipc_2014_districts
}

print("Historical normalized districts:", len(historical_normalized))
print("2014 normalized districts:", len(ipc_2014_normalized))

print("\nPresent historically but not in 2014:")
print(sorted(historical_normalized - ipc_2014_normalized))

print("\nPresent in 2014 but not historically:")
print(sorted(ipc_2014_normalized - historical_normalized))
```

```python
# Prepare Maharashtra IPC and women-crime datasets for merging

# Copy the filtered Maharashtra datasets
ipc_historical_mh = ipc_mh.copy()
women_historical_mh = women_mh.copy()

# Normalize district names
ipc_historical_mh["District_Normalized"] = (
    ipc_historical_mh["District"].apply(normalize_district_name)
)

women_historical_mh["District_Normalized"] = (
    women_historical_mh["District"].apply(normalize_district_name)
)

# Create a consistent merge key
merge_columns = [
    "State_UT",
    "District_Normalized",
    "Year"
]

# Merge IPC + women-crime data
historical_merged = pd.merge(
    ipc_historical_mh,
    women_historical_mh,
    on=merge_columns,
    how="inner",
    suffixes=("_ipc", "_women"),
    validate="one_to_one"
)

print("IPC Maharashtra rows:", len(ipc_historical_mh))
print("Women-crime Maharashtra rows:", len(women_historical_mh))
print("Merged rows:", len(historical_merged))

print("\nMerged shape:", historical_merged.shape)

print("\nFirst 5 rows:")
display(historical_merged.head())
```

```python
# ============================================================
# STEP 6 — FEATURE ENGINEERING FOR ML
# ============================================================

import numpy as np

df = maharashtra_crime_2001_2014.copy()

# ------------------------------------------------------------
# 1. Identify numeric crime columns
# ------------------------------------------------------------

id_columns = ["State_UT", "District", "Year"]

crime_columns = [
    c for c in df.columns
    if c not in id_columns
]

df[crime_columns] = df[crime_columns].apply(
    pd.to_numeric,
    errors="coerce"
).fillna(0)


# ------------------------------------------------------------
# 2. Women-safety aggregate features
# ------------------------------------------------------------

df["WOMEN_CRIME_TOTAL"] = (
    df["RAPE"]
    + df["Kidnapping & Abduction of Women"]
    + df["DOWRY DEATHS"]
    + df["ASSAULT ON WOMEN WITH INTENT TO OUTRAGE HER MODESTY"]
    + df["INSULT TO MODESTY OF WOMEN"]
    + df["CRUELTY BY HUSBAND OR HIS RELATIVES"]
    + df["IMPORTATION OF GIRLS FROM FOREIGN COUNTRIES"]
)


# ------------------------------------------------------------
# 3. Violent-crime aggregate
# ------------------------------------------------------------

df["VIOLENT_CRIME_TOTAL"] = (
    df["MURDER"]
    + df["ATTEMPT TO MURDER"]
    + df["CULPABLE HOMICIDE NOT AMOUNTING TO MURDER"]
    + df["RAPE"]
    + df["Total Kidnapping & Abduction"]
    + df["ROBBERY"]
    + df["RIOTS"]
    + df["HURT/GREVIOUS HURT"]
)


# ------------------------------------------------------------
# 4. Property-crime aggregate
# ------------------------------------------------------------

df["PROPERTY_CRIME_TOTAL"] = (
    df["BURGLARY"]
    + df["THEFT"]
    + df["AUTO THEFT"]
    + df["OTHER THEFT"]
)


# ------------------------------------------------------------
# 5. Create previous-year features
# ------------------------------------------------------------

df = df.sort_values(
    ["District", "Year"]
).reset_index(drop=True)

group = df.groupby("District", group_keys=False)

df["IPC_TOTAL_LAG_1"] = group["TOTAL IPC CRIMES"].shift(1)

df["WOMEN_CRIME_LAG_1"] = group["WOMEN_CRIME_TOTAL"].shift(1)

df["VIOLENT_CRIME_LAG_1"] = group["VIOLENT_CRIME_TOTAL"].shift(1)


# ------------------------------------------------------------
# 6. Rolling historical averages
#    IMPORTANT: shift first to avoid data leakage
# ------------------------------------------------------------

df["IPC_ROLLING_MEAN_3"] = (
    group["TOTAL IPC CRIMES"]
    .transform(
        lambda x: x.shift(1).rolling(3, min_periods=1).mean()
    )
)

df["WOMEN_ROLLING_MEAN_3"] = (
    group["WOMEN_CRIME_TOTAL"]
    .transform(
        lambda x: x.shift(1).rolling(3, min_periods=1).mean()
    )
)


# ------------------------------------------------------------
# 7. Year-over-year IPC trend
# ------------------------------------------------------------

df["IPC_YEARLY_CHANGE"] = (
    df["TOTAL IPC CRIMES"]
    - df["IPC_TOTAL_LAG_1"]
)

df["WOMEN_YEARLY_CHANGE"] = (
    df["WOMEN_CRIME_TOTAL"]
    - df["WOMEN_CRIME_LAG_1"]
)


# ------------------------------------------------------------
# 8. Create NEXT-YEAR prediction targets
# ------------------------------------------------------------

df["TARGET_NEXT_YEAR_IPC"] = (
    group["TOTAL IPC CRIMES"].shift(-1)
)

df["TARGET_NEXT_YEAR_WOMEN_CRIME"] = (
    group["WOMEN_CRIME_TOTAL"].shift(-1)
)


# ------------------------------------------------------------
# 9. Remove rows where next-year target doesn't exist
# ------------------------------------------------------------

modeling_df = df.dropna(
    subset=[
        "TARGET_NEXT_YEAR_IPC",
        "TARGET_NEXT_YEAR_WOMEN_CRIME"
    ]
).copy()


# ------------------------------------------------------------
# 10. Final validation
# ------------------------------------------------------------

print("=" * 70)
print("ML DATASET")
print("=" * 70)

print("\nOriginal rows:", len(df))
print("Modeling rows:", len(modeling_df))
print("Feature columns:", len(modeling_df.columns))

print("\nYear range:")
print(
    modeling_df["Year"].min(),
    "→",
    modeling_df["Year"].max()
)

print("\nTarget availability:")
print(
    modeling_df[
        [
            "TARGET_NEXT_YEAR_IPC",
            "TARGET_NEXT_YEAR_WOMEN_CRIME"
        ]
    ].isna().sum()
)

print("\nMissing values:")
missing = modeling_df.isna().sum()
print(missing[missing > 0])

print("\nFeature preview:")
display(
    modeling_df[
        [
            "District",
            "Year",
            "TOTAL IPC CRIMES",
            "WOMEN_CRIME_TOTAL",
            "VIOLENT_CRIME_TOTAL",
            "PROPERTY_CRIME_TOTAL",
            "IPC_TOTAL_LAG_1",
            "IPC_ROLLING_MEAN_3",
            "TARGET_NEXT_YEAR_IPC",
            "TARGET_NEXT_YEAR_WOMEN_CRIME"
        ]
    ].head(15)
)
```

```python
# ============================================================
# STEP 7 — CHECK YEAR/DISTRICT CONTINUITY
# ============================================================

# Show number of records by year
print("Rows by year:")
print(
    df.groupby("Year")
      .size()
      .sort_index()
)

print("\n" + "=" * 70)

# Check which district-year combinations are missing
years = sorted(df["Year"].unique())

expected_years = set(range(2001, 2015))

district_year_check = (
    df.groupby("District")["Year"]
      .apply(set)
)

print("District continuity problems:\n")

for district, available_years in district_year_check.items():

    missing_years = sorted(
        expected_years - available_years
    )

    if missing_years:
        print(
            f"{district}: missing {missing_years}"
        )

print("\n" + "=" * 70)

# Check how many rows have a valid next-year target
print("Rows with next-year target:",
      df["TARGET_NEXT_YEAR_IPC"].notna().sum())

print("Rows without next-year target:",
      df["TARGET_NEXT_YEAR_IPC"].isna().sum())

print("\nRows without target:")
display(
    df[
        df["TARGET_NEXT_YEAR_IPC"].isna()
    ][
        ["District", "Year"]
    ].sort_values(["Year", "District"])
)
```

```python
# ============================================================
# STEP 8 — FINAL LEAKAGE-SAFE MODELING DATASET
# ============================================================

# Start from the engineered dataset
modeling_df = df.copy()

# ------------------------------------------------------------
# Keep only rows that have:
# 1. A previous-year observation
# 2. A next-year target
# ------------------------------------------------------------

required_columns = [
    "IPC_TOTAL_LAG_1",
    "WOMEN_CRIME_LAG_1",
    "VIOLENT_CRIME_LAG_1",
    "IPC_ROLLING_MEAN_3",
    "WOMEN_ROLLING_MEAN_3",
    "TARGET_NEXT_YEAR_IPC",
    "TARGET_NEXT_YEAR_WOMEN_CRIME"
]

modeling_df = modeling_df.dropna(
    subset=required_columns
).copy()


# ------------------------------------------------------------
# We cannot predict beyond 2013 because 2015 data doesn't exist
# ------------------------------------------------------------

modeling_df = modeling_df[
    modeling_df["Year"] < 2014
].copy()


# ------------------------------------------------------------
# Sort properly
# ------------------------------------------------------------

modeling_df = (
    modeling_df
    .sort_values(["Year", "District"])
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

print("=" * 70)
print("FINAL MODELING DATASET")
print("=" * 70)

print("\nShape:", modeling_df.shape)

print("\nYear coverage:")
print(
    modeling_df["Year"]
    .value_counts()
    .sort_index()
)

print("\nDistrict count:")
print(
    modeling_df["District"].nunique()
)

print("\nMissing values:")
missing = modeling_df.isna().sum()
print(missing[missing > 0])

print("\nDuplicate District-Year records:")
print(
    modeling_df.duplicated(
        subset=["District", "Year"]
    ).sum()
)

print("\nTarget statistics:")

print("\nNext-year IPC:")
print(
    modeling_df["TARGET_NEXT_YEAR_IPC"].describe()
)

print("\nNext-year women crime:")
print(
    modeling_df["TARGET_NEXT_YEAR_WOMEN_CRIME"].describe()
)

print("\nFirst 10 modeling rows:")
display(
    modeling_df[
        [
            "District",
            "Year",
            "TOTAL IPC CRIMES",
            "WOMEN_CRIME_TOTAL",
            "VIOLENT_CRIME_TOTAL",
            "PROPERTY_CRIME_TOTAL",
            "IPC_TOTAL_LAG_1",
            "IPC_ROLLING_MEAN_3",
            "TARGET_NEXT_YEAR_IPC",
            "TARGET_NEXT_YEAR_WOMEN_CRIME"
        ]
    ].head(10)
)
```
